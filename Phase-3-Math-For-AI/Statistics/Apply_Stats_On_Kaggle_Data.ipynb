```python
# 10. Percentiles
print(df["SalePrice"].quantile(0.95))
```

```python
highPrice = df[df["SalePrice"] > df["SalePrice"].quantile(0.95)]
print(highPrice.SalePrice)
```

```python
# 11. IQR
q1 = df["SalePrice"].quantile(0.25)
q3 = df["SalePrice"].quantile(0.75)
IQR = q3 - q1

lower = q1 - (1.5*IQR)
upper = q3 + (1.5*IQR)

df = df[(df["SalePrice"] >= lower) & (df["SalePrice"] <= upper)]
print(df)
```
