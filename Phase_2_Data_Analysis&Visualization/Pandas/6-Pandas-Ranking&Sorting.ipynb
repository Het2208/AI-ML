```python
import pandas as pd
```

```python
df = pd.read_csv("pokemon_150.csv")
```

```python
# sort_values() — Sort by Column Values
# Sort Ascending (Default)
print(df.sort_values(by = "Weight"))
```

```python
# Sort Descending
print(df.sort_values(by = "Weight" , ascending=False))
```

```python
# Sort by Multiple Columns
print(df.sort_values(by = ["Weight" , "Height"])) # First sort by Weight , If Weight are same → sorts by Height.
```

```python
# Different order for different columns
print(df.sort_values(by = ["Weight" , "Height"] , ascending = [True , False]))
```

```python
# Permanent Sorting (inplace=True)
# Normally sorting creates new dataframe.
df.sort_values(by = "Weight" , inplace = True)
```

```python
print(df)
# Original dataframe changes permanently.
```

```python
# Reset Index After Sorting
# After sorting index becomes messy.
df.sort_values(by = "Weight").reset_index(drop=True)
```

```python
# Sort by Index — sort_index()
# Sorts using row labels/index.
df.sort_index()
```

```python
# Sorting Strings
# Sorts alphabetically.
df.sort_values(by="Name")
```

```python
# Sorting Missing Values (NaN)
# Should missing values appear at the top?
# Or at the bottom?
df.sort_values(by = "Type2" , na_position="first")
```

```python
df.sort_values(by = "Type2" , na_position="last")
```

```python
# Largest & Smallest Values
df.nlargest(3, "Height")
```

```python
df.nsmallest(3, "Height")
```

```python
# Highest Rank First
df2 = pd.read_csv("pokemon_150.csv")
df2["Weight"].rank(ascending=False)
```

```python
# Ranking Methods
# When duplicate values exist:
```

```python
# Method 1 — average (Default)
df2["Rank"] = df2["Height"].rank(method="average" , ascending = False)
print(df2.to_string())
```

```python
# Method 2 — Min Method
df2["Rank"] = df2["Height"].rank(method="min" , ascending = False)
print(df2.to_string())
```

```python
# Method 2 — Max Method
df2["Rank"] = df2["Height"].rank(method="max" , ascending = False)
print(df2.to_string())
```

```python
# Method 4 — first
# Ranks based on appearance order.
df2["Rank"] = df2["Height"].rank(method="first" , ascending = False)
print(df2.to_string())
```

```python
# Method 5 — dense
# Main Rule of Dense Ranking
# Duplicate values get same rank
# AND
# Next rank increases by only 1
# No gaps.
    
df2["Rank"] = df2["Height"].rank(method="dense" , ascending = False)
print(df2.to_string())
```

```python
# Percentage Ranking (pct=True) for statistic
df["Weight"].rank(pct = True)
```

```python
# Percentage Ranking (pct=True)
df2["TypeRank"] = (df.groupby("Type1")["Weight"].rank(ascending=False))
print(df2.to_string())
```
