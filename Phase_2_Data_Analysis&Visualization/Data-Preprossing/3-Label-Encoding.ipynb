```python
# Label Encoding is a technique used to convert:
# Text/Categorical values → Numbers
# Because Machine Learning models understand:
# Numbers
# not text.

# Why Label Encoding is Needed
# Suppose dataset contains:
# ShirtSize
# Small
# Medium
# Large

# Machine Learning cannot directly understand:

# "Small"
# "Medium"
# "Large"

# So we convert them into numbers.

# After Label Encoding
# ShirtSize	Encoded
# Small	       2
# Medium	       1
# Large	       0

# Now ML models can use the data.

# Important Understanding ⭐

# Label encoding:

# Assigns a unique number to each category

# Example:

# Apple  → 0
# Banana → 1
# Mango  → 2
# Real-Life Analogy

# Imagine school IDs:

# Student Name	Roll Number
# Het	1
# Raj	2
# Jay	3

# Roll number is just an encoded representation.

# Same idea.

# Example Dataset (NOT CITY)

# Let's use:

# Customer Membership Type
# Original Dataset
# Customer	Membership
# A	Gold
# B	Silver
# C	Platinum
# D	Gold
# E	Silver

# Goal : Convert Membership into numbers.
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.DataFrame({"Student": ["A", "B", "C", "D", "E"],"CourseLevel": ["Beginner","Advanced","Intermediate","Beginner","Advanced"]})

print(df)
```

```python
# Createing LabelEncoder Object
le = LabelEncoder()
```

```python
# Apply Label Encoding
df["Encoded_Level"] = le.fit_transform(df["CourseLevel"])


# PART A — fit()

# Encoder studies categories.
# It learns:

# Advanced
# Beginner
# Intermediate
    
# Stores them internally.

# PART B — transform()

# Now conversion happens:

# Advanced     → 0
# Beginner     → 1
# Intermediate → 2
```

```python
print(df)

# WHY DID ADVANCED BECOME 0?

# Because:

# LabelEncoder sorts alphabetically

# Order:

# Advanced
# Beginner
# Intermediate

# Then numbering starts from 0.
```

```python
# Understanding classes_

print(le.classes_)

# which category got which number
# Position index = encoded value.
```

```python
# Reverse Transformation

# Suppose model predicts:2
# How to know actual category?
# Use:inverse_transform()

le.inverse_transform([2])
```

```python
# IMPORTANT PROBLEM IN LABEL ENCODING ⚠️

# This is the MOST IMPORTANT theoretical concept.

# Problem: False Ordering

# Suppose:

# Advanced     → 0
# Beginner     → 1
# Intermediate → 2

# Some ML models may think:

# 2 > 1 > 0

# Meaning:

# Intermediate > Beginner > Advanced

# Mathematically it assumes order exists.

# But in reality:

# these are just names

# This creates:

# False Relationship Problem
# Which Models Get Affected?

# Mostly:

# Linear Regression
# Logistic Regression
# KNN
# SVM

# because they use:

# distances
# magnitudes
# mathematical relationships
# Which Models Usually Don't Care Much?

# Tree-based models:

# Decision Trees
# Random Forest
# XGBoost

# less affected.

# WHEN SHOULD WE USE LABEL ENCODING? ⭐

# Very important interview question.

# GOOD USE CASE — Ordinal Data

# Ordinal means:

# categories have meaningful order
# Example — Shirt Size
# Size
# Small
# Medium
# Large

# Clearly:

# Small < Medium < Large

# So encoding:

# Small  → 0
# Medium → 1
# Large  → 2

# makes logical sense.

# Another Example

# Education level:

# School
# College
# Masters
# PhD

# Order exists.

# Label encoding works well.

# BAD USE CASE — Nominal Data ❌

# Nominal means:

# No meaningful order exists
# Example
# Fruit
# Apple
# Mango
# Banana

# Encoding:

# Apple  → 0
# Banana → 1
# Mango  → 2

# Problem:

# Mango > Banana > Apple

# No real meaning.

# Better approach:

# One-Hot Encoding
```

```python
# MANUAL LABEL ENCODING
df["CourseLevel"] = df["CourseLevel"].map({
    "Beginner" : 0,
    "Intermediate" : 1,
    "Advanced" : 2
})

print(df)

# WHY MANUAL MAPPING IS BETTER SOMETIMES
# Because:YOU control the order
# instead of alphabetical ordering.
# Very important for ordinal data.
```

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Department": ["HR", "IT", "Finance", "HR"],
    "Education": ["Graduate", "PhD", "Masters", "Graduate"]
})

print(df)
```

```python
# WRONG BEGINNER THINKING
le = LabelEncoder()

df["Department"] = le.fit_transform(df["Department"])
df["Education"] = le.fit_transform(df["Education"])

print(df)


# Why This Is Dangerous
# Because:Every column has DIFFERENT categories.
    
# Department Categories
# HR
# IT
# Finance

# Education Categories
# Graduate
# Masters
# PhD

# These are completely different meanings.

# So:Each column needs its OWN category mapping.

```

```python
# Step 1 — Encode Department
# from sklearn.preprocessing import LabelEncoder

# le = LabelEncoder()

# df["Department"] = le.fit_transform(df["Department"])
# Internally Stored Mapping
# print(le.classes_)

# Output:

# ['Finance' 'HR' 'IT']

# Meaning:

# Finance → 0
# HR      → 1
# IT      → 2
# Step 2 — Encode Education Using SAME Encoder
# df["Education"] = le.fit_transform(df["Education"])

# Now check:

# print(le.classes_)

# Output:

# ['Graduate' 'Masters' 'PhD']
# What Happened? ⚠️

# Old Department mapping:

# Finance → 0
# HR      → 1
# IT      → 2

# is LOST.

# Now encoder stores:

# Graduate → 0
# Masters  → 1
# PhD      → 2
# Why Is This a Problem?

# Suppose later:

# le.inverse_transform([1])

# You expect:

# HR

# But now result becomes:

# Masters

# because mapping changed.

# THIS IS WHY SEPARATE ENCODERS ARE USED ⭐⭐⭐

# Each column needs:

# its own independent mapping memory
```

```python
df2 = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Department": ["HR", "IT", "Finance", "HR"],
    "Education": ["Graduate", "PhD", "Masters", "Graduate"]
})

dept_encoder = LabelEncoder()
edu_encoder = LabelEncoder()

df["Department"] = dept_encoder.fit_transform(df2["Department"])

df["Education"] = edu_encoder.fit_transform(df2["Education"])

print(df2)
```

```python
# 1. Handling Unseen Labels 

# MOST IMPORTANT remaining topic.

# Problem

# Suppose model trained on:

# HR
# IT
# Finance

# Encoder learned:

# Finance → 0
# HR → 1
# IT → 2
# Later New Data Comes
# Marketing

# Encoder never saw this category before.

# Example

# from sklearn.preprocessing import LabelEncoder
# le = LabelEncoder()
# train = ["HR", "IT", "Finance"]
# le.fit(train)
# le.transform(["Marketing"])

# ERROR ⚠️
# ValueError:
# y contains previously unseen labels
# WHY THIS HAPPENS

# Because encoder only knows:

# Finance
# HR
# IT

# No mapping exists for:

# Marketing
```

```python
dept = LabelEncoder()


train = ["HR", "IT", "Finance"]

dept.fit(train)

dept.transform(["Marketing"])
```

```python
# Solution 1 — Use map()
mapping = {
    "HR": 0,
    "IT": 1,
    "Finance": 2
}

df["Department"] = df["Department"].map(mapping)
```

```python
# Solution 2 — Add Unknown Category
mapping = {
    "HR": 0,
    "IT": 1,
    "Finance": 2,
    "Unknown": -1
}
```

```python
# OrdinalEncoder
# Used professionally instead of multiple LabelEncoders.

from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder()

df[["Department", "Education"]] = oe.fit_transform(
    df[["Department", "Education"]]
)
```
