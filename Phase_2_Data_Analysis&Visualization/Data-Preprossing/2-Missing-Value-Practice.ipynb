```python
import pandas as pd
```

```python
df = pd.read_csv("employees_missing_data.csv")
print(df)
```

```python
# 1. Check Missing Values
df.isna().sum()
```

```python
# 2. Drop Rows if Very Few Missing
df.dropna()
```

```python
# 3. Fill Numerical Data with Median
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
print(df)
```

```python
# 4. Fill Categorical Data with Mode
df["City"] = df["City"].fillna(df["City"].mode()[0])
print(df)
```

```python
# 5. Fill Age with Mean
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df)
```

```python
# 6. Group-Based Filling
# Fill salary based on department average.
df["Salary"] = (df.groupby("Department")["Salary"].transform(lambda x : x.fillna(x.mean())))
print(df)
```

```python
# 7. Forward Fill
df.fillna(method="ffill")
```

```python
# 8. Backward Fill
df.fillna(method="bfill")
```

```python
# 9. Remove Mostly Empty Columns
(df.isna().sum() / len(df)) * 100
```
