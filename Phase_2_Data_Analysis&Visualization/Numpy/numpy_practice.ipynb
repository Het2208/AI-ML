## 🧠 Problem 1: Normalize Data (VERY IMPORTANT)
**Formula:** `(x - mean) / std`

```python
# Input
arr = np.array([10, 20, 30, 40, 50])

# Task: Normalize using (x - mean) / std

# YOUR CODE HERE
normalized = (arr - arr.mean()) / arr.std()

print(normalized)

# Expected Output (approx):
# [-1.414, -0.707,  0.0,   0.707,  1.414]
```

## 🧠 Problem 2: Feature Filtering
**Task:** Get only values greater than 10

```python
# Input
arr = np.array([5, 12, 7, 18, 3])

# Task: Get values > 10

# YOUR CODE HERE
filtered = arr[arr>10]

print(filtered)

# Expected Output:
# [12, 18]
```

## 🧠 Problem 3: Replace Outliers
**Task:** Replace values > 10 with the mean of the entire array

```python
# Input
arr = np.array([1, 2, 100, 3, 4])

# Task: Replace values > 10 with mean of the array

# YOUR CODE HERE
mean_val = arr.mean()
arr_replaced = np.where(arr > 10 , mean_val , arr)

print(arr_replaced)

# Expected Output:
# [ 1  2  22  3  4]
```

## 🧠 Problem 4: Dot Product (ML Core)
**Task:** Compute the dot product of two vectors

```python
# Input
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Task: Compute dot product

# YOUR CODE HERE
dot = np.dot(a,b)

print(dot)

# Expected Output:
# 32
```

## 🧠 Problem 5: Matrix Multiplication
**Task:** Multiply two 2x2 matrices using `@` operator

```python
# Input
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Task: Compute A @ B

# YOUR CODE HERE
result = np.matmul(A,B)

print(result)

# Expected Output:
# [[19, 22],
#  [43, 50]]
```

## 🧠 Problem 6: Column-wise Mean
**Task:** Find the mean of each column in a 2D array

```python
# Input
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

# Task: Find mean of each column (axis=0)

# YOUR CODE HERE
col_means = arr.mean(axis = 0)

print(col_means)

# Expected Output:
# [2.5, 3.5, 4.5]
```

## 🧠 Problem 7: One-Hot Encoding (Basic ML)
**Task:** Convert class labels to one-hot encoded matrix

```python
# Input
arr = np.array([0, 1, 2])

# Task: Convert to one-hot encoding

# YOUR CODE HERE
one_hot = np.eye(3)[arr]

print(one_hot)

# Expected Output:
# [[1, 0, 0],
#  [0, 1, 0],
#  [0, 0, 1]]
```

## 🧠 Problem 8: Find Index of Max Value per Column
**Task:** Return the row index of the maximum value in each column

```python
# Input
arr = np.array([[10, 20, 30],
                [40, 15, 25]])

# Task: Find index of max value in each column (axis=0)

# YOUR CODE HERE
max_indices = arr.argmax(axis = 0)

print(max_indices)

# Expected Output:
# [1, 0, 0]
```

## 🧠 Problem 9: Standard Deviation & Variance
**Task:** Compute both std and variance of the array

```python
# Input
arr = np.array([2, 4, 6, 8])

# Task: Find standard deviation and variance

# YOUR CODE HERE
std = arr.std()
var = arr.var()

print(f"std = {std:.3f}")
print(f"var = {var:.1f}")

# Expected Output:
# std = 2.236
# var = 5.0
```

## 🧠 Problem 10: Apply Condition (np.where)
**Task:** Replace values greater than 5 with 100 using `np.where`

```python
# Input
arr = np.array([1, 6, 3, 8])

# Task: Replace values > 5 with 100

# YOUR CODE HERE
result = np.where(arr > 5 , 100 , arr)

print(result)

# Expected Output:
# [  1, 100,   3, 100]
```
