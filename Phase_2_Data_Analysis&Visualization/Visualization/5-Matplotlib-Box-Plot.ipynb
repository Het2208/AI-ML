```python
# ---------------------------------------------------------
# BOXPLOT
# ---------------------------------------------------------

# • Boxplots provide a means of depicting groups of numbers
#   through their quartiles.

# • Quartiles means three points dividing a group into
#   four equal parts.

# • In boxplot, data will be divided in 4 parts using:
#       Q1 -> 25th Percentile
#       Q2 -> 50th Percentile (Median)
#       Q3 -> 75th Percentile

# ---------------------------------------------------------
# IMPORTANT TERMS
# ---------------------------------------------------------

# Q1  = First Quartile  (25th Percentile)
# Q2  = Median          (50th Percentile)
# Q3  = Third Quartile  (75th Percentile)

# IQR = Interquartile Range
#       IQR = Q3 - Q1

# Minimum Value:
#       Q1 - 1.5 * IQR

# Maximum Value:
#       Q3 + 1.5 * IQR

# Whiskers:
#       Lines extending from the box showing spread of data.

# Outliers:
#       Data points lying outside the whisker range.

# ---------------------------------------------------------
# VISUAL STRUCTURE OF BOXPLOT
# ---------------------------------------------------------

# Outliers | Whisker | Q1 | Median | Q3 | Whisker | Outliers

# ---------------------------------------------------------
# PURPOSE OF BOXPLOT
# ---------------------------------------------------------

# • Shows data distribution
# • Detects outliers
# • Displays spread of data
# • Helps compare multiple datasets
# • Easy visualization of quartiles and median
```

```python
import matplotlib.pyplot as plt

data = [10, 12, 14, 15, 18, 20, 22, 25, 100]

plt.boxplot(data)

plt.title("Boxplot Example")
plt.show()
```

```python
plt.boxplot(data , notch=True)
plt.show()
```

```python
plt.boxplot(data , widths=0.2)
plt.show()
```

```python
plt.boxplot(data , vert=False)
plt.show()
```
