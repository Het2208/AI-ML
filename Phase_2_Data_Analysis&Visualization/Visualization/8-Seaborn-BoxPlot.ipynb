```python
# BOXPLOT

# Boxplot is a graph that summarizes:
# entire distribution of data

# using:
# box
# lines
# dots


# BOXPLOT SHOWS
# Minimum
# Maximum
# Median
# Quartiles
# Outliers

# All in ONE graph.
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({"Salary": [25000,30000,35000,40000,45000,500000]})
print(df)
```

```python
sns.boxplot(data=df["Salary"])
plt.show()
```

```python
# 1. Median Line
# Middle line inside box.
# middle value of data

# 50% values above it.
# 50% below it.
```

```python
# 2. Box 

# Represents:
# middle 50% of data

# called:
# Interquartile Range (IQR)
```

```python
# WHAT IS IQR?

# IQR measures:
# spread of middle data

# Formula:
# IQR=Q3−Q1
```

```python
# Term	Meaning
# Q1	25% data point
# Q2	Median
# Q3	75% data point
```

```python
# |----whisker----[ BOX ]----whisker----|
#                    |
#                 median
```

```python
# 3. Whiskers 

# Lines extending from box.

# Represent:
# normal minimum and maximum range
```

```python
# 4. Outliers 

# Dots outside whiskers.

# These are:
# unusually extreme values

# Example:
# 500000
# salary among normal salaries.
```

```python
# HOW OUTLIERS DETECTED? 

# Boxplot uses:
# Lower=Q1−1.5(IQR)

# and

# Upper=Q3+1.5(IQR)

# Values outside this range:
# become outliers
```

```python
# CUSTOMIZING BOXPLOT
```

```python
sns.boxplot(
    data=df["Salary"],
    color="orange"
)
```

```python
sns.boxplot(x=df["Salary"]) # Horizontal Boxplot
```

```python
sns.boxplot(y=df["Salary"]) # Vertical Boxplot
```

```python
sns.boxplot(data=df["Salary"])
plt.title("Salary Distribution") # Add Title
plt.show() 
```

```python
# BOXPLOT WITH CATEGORIES 
```

```python
df = pd.DataFrame({"Department" : ["HR" , "HR" , "IT" , "IT"] , "Salary" : [30000 , 35000 , 70000 , 90000]})
print(df)
```

```python
sns.boxplot(x="Department" , y="Salary" , data=df)
plt.show()
```

```python
# BOXPLOT INTERPRETATION

# Narrow Box
# low variation

# Wide Box
# high variation

# Many Outliers
# data has extreme values

# Median Not Centered
# skewed distribution
```

```python
# Use Boxplot When:
# detecting outliers
# quick statistical summary
# simpler interpretation

# Boxplot Used For
# salary analysis
# outlier detection
# data cleaning
# feature analysis
```
