```python
# If one variable changes,
# does another variable also change?
```

```python
hours = [1, 2, 3, 4, 5]
marks = [20, 35, 50, 65, 80]

plt.scatter(hours, marks , color="red" , s=100 , marker="^" , alpha=0.5) # alpha -> Add Transparency Useful when:many points overlap
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
maths = [20 , 35 , 50 , 65 , 80]
science = [10 , 40 , 55 , 60 , 85]

plt.scatter(hours, maths, label="Maths")
plt.scatter(hours, science, label="Science")

plt.legend()

plt.show()
```
