```python
import matplotlib.pyplot as plt
import random
```

```python
runs = [random.randint(0,36) for _ in range(1 , 21)]
over = range(1,21)

plt.bar(over , runs , color="orange" , width=0.7)
plt.title("Run per over")
plt.xlabel("Overs")
plt.ylabel("Runs")
plt.show()
```

```python
plt.barh(over , runs)
plt.show()
```

```python
bars = plt.bar(over,runs)
plt.bar_label(bars)
plt.show()
```
