```python
import matplotlib.pyplot as plt
```

```python
# PART 1 — FIGURE SIZING -> size of the entire graph window

# WHY IMPORTANT?

# Default graph size may:
# look too small
# overlap labels
# hide information
# become unreadable

# So we control:
# width and height
```

```python
# plt.figure(figsize=(width, height)) width&height in inches 

# COMMON FIGURE SIZES 
# Figure Size	Usage
# (6,4)	        Default/simple
# (8,5)	        Most common
# (12,6)	    Dashboards
# (15,8)	    Large reports
```

```python
months = ["Jan", "Feb", "Mar"]
sales = [100, 150, 120]

plt.figure(figsize = (6,4))

plt.plot(months , sales)
plt.show()
```

```python
plt.figure(figsize = (8,5))

plt.plot(months , sales)
plt.show()
```

```python
# PART 2 — SUBPLOTS  ->  multiple graphs inside one figure

# That single page = figure

# Each individual graph = subplot
```

```python
# WHY SUBPLOTS IMPORTANT?

# Used when:

# comparing graphs
# creating dashboards
# showing multiple analyses together
# avoiding many separate windows
```

```python
# plt.subplot(rows, columns, position)

# Parameter	 Meaning
# rows	     Total rows
# columns	 Total columns
# position	 Current graph position
```

```python
x = [1,2,3]
y1 = [10,20,30]
y2 = [30,20,10]

plt.figure(figsize=(10,5))

# 1st graph
plt.subplot(1,2,1)
plt.plot(x, y1)
plt.title("Graph 1")

# 2nd graph
plt.subplot(1,2,2)
plt.plot(x, y2)
plt.title("Graph 2")

plt.show()


# Created:

# 1 row
# 2 columns

# Layout:

# [Graph1] [Graph2]
```

```python
# MIXING DIFFERENT CHART TYPES

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot([1,2,3],[4,5,6])
plt.title("Line")

plt.subplot(2,2,2)
plt.bar(["A","B","C"],[5,7,3])
plt.title("Bar")

plt.subplot(2,2,3)
plt.scatter([1,2,3],[7,2,5])
plt.title("Scatter")

plt.subplot(2,2,4)
plt.pie([40,30,30])
plt.title("Pie")

plt.show()
```

```python
# Problem

# Subplots may overlap:

# titles
# labels
# axes

# Solution
# plt.tight_layout()

# Automatically adjusts spacing.

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot([1,2,3],[4,5,6])
plt.title("Line")

plt.subplot(2,2,2)
plt.bar(["A","B","C"],[5,7,3])
plt.title("Bar")

plt.subplot(2,2,3)
plt.scatter([1,2,3],[7,2,5])
plt.title("Scatter")

plt.subplot(2,2,4)
plt.pie([40,30,30])
plt.title("Pie")

plt.tight_layout()

plt.show()
```
