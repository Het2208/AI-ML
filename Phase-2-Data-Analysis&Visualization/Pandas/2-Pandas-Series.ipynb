```python
import pandas as pd
```

```python
# Series is one-dimensional labeled array that can hold any datatype 
# think of it like a single column in spredsheets (1 - Dimensional)

# pd.Series([] , index=[])  -> constructor is used for series
```

```python
series1 = pd.Series([100 , 200 , 300])
print(series1)
```

```python
series2 = pd.Series(["Het" , "Meet" , "Preet"])
print(series2)
```

```python
series3 = pd.Series([True , False , True])
print(series3)
```

```python
series4 = pd.Series([101.1 , 102.2 , 103.3] , index=["a" , "b" , "c"])
print(series4)

print("****************************************************************************************************")
print(series4.loc["a"]) # loc -> location by label (loc = property)
print("****************************************************************************************************")

series4.loc["c"] = 200.0
print(series4)

print("****************************************************************************************************")

print(series4.iloc[0]) # iloc -> integer location 
print(series4.iloc[1])
```

```python
print(series4[series4 >= 200])
```

```python
dictSeries = pd.Series({"Day 1" : 1700 , "Day 2" : 2150 , "Day 3" : 1900 , "Day 4" : 2000})
print(dictSeries)
```

```python
print(dictSeries.loc["Day 2"])
print(dictSeries.loc["Day 4"])

print("****************************************************************************************************")

dictSeries.loc["Day 1"] = 1850

print(dictSeries)

print("****************************************************************************************************")

print(dictSeries.iloc[0])
print(dictSeries.iloc[3])

print("****************************************************************************************************")

print(dictSeries[dictSeries >= 2000])
```
