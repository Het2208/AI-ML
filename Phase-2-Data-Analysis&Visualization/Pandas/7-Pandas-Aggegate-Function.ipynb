```python
# Single column aggergation
print(df["Height"].mean())
```

```python
print(df["Height"].mode())
```

```python
print(df["Height"].sum())
```

```python
print(df["Height"].count())
```

```python
print(df["Height"].max())
```

```python
# groupby()
group = df.groupby("Type1")
print(group) # <pandas.core.groupby.generic.DataFrameGroupBy object at 0x0000028B75523230>
```

```python
print(group["Height"].count())
```

```python
print(group["Height"].sum())
```

```python
print(group["Height"].mean())
```

```python
print(group["Height"].median())
```

```python
print(group["Weight"].max())
```

```python
print(group["Weight"].min())
```

```python
# Multiple Aggregations using .agg()
print(group["Height"].agg(["count" , "mean" , "max" , "min"]))

print("*"*100)

# Custom aggregation names
print(group["Height"].agg(Total = "count" , Average = "mean" , Maximum = "max" ,Minimum = "min"))
```

```python
# Grouping by Multiple Columns
group2 = df.groupby(["Type1" , "Type2"])
print(group2["Height"].count().to_string())
```

```python
# size() vs count()
print(group["Type2"].count()) # Counts NON-null values.

print("*"*100)

print(group["Type2"].size()) # Counts total rows including nulls.
```

```python
# # value_counts()
# Useful for:
# Frequency analysis
# Category counting
# EDA (Exploratory Data Analysis)

print(df["Type1"].value_counts())
```

```python
# Resetting Index after GroupBy
# GroupBy often creates index problems.
# Very important for clean tables.

print(group["Height"].mean())

print("*"*100)

print(group["Height"].mean().reset_index())
```

```python
# Sorting Grouped Results
print(group["Height"].count().sort_values())
```

```python
print(group["Height"].count().sort_values(ascending=False))
```

```python
# get_group()
# Extract one specific group.

print(group.get_group("Fire"))
```

```python
# Multiple Column Aggregation
# You currently aggregate only one column at a time.
print(group[["Height", "Weight"]].mean())
```

```python
# Different Aggregation for Different Columns
print(group.agg({
    "Height": "mean",
    "Weight": "max"
}))
```

```python
# 12. transform() (Advanced but Important)

# Returns same size output.

# Example:

# df["AvgHeightByType"] = df.groupby("Type1")["Height"].transform("mean")

# Adds group statistics back to original dataframe.

# Very important in feature engineering.
```

```python
# filter() in GroupBy
print(df.groupby("Type1").filter(lambda x : len(x) > 20))
```

```python
# nunique()
# Counts unique values.

print(group["Height"].nunique())
print("*"*100)
print(group["Height"].count())
```

```python
# 1. as_index=False

# You used reset_index(), but this is another cleaner approach.

# Example:

# df.groupby("Type1", as_index=False)["Height"].mean()

# Instead of:

# df.groupby("Type1")["Height"].mean().reset_index()

# Both similar.

# 2. Named Aggregation (Modern Style)

# You probably used:

# group.agg(["mean", "max"])

# But modern Pandas style is:

# df.groupby("Type1").agg(
#     AvgHeight=("Height", "mean"),
#     MaxWeight=("Weight", "max")
# )

# Very important in professional projects.

# 3. pivot_table() ⭐

# This is the biggest important topic remaining.

# It is basically advanced groupby.

# Example:

# pd.pivot_table(
#     df,
#     values="Height",
#     index="Type1",
#     aggfunc="mean"
# )

# Multi-dimensional summaries become very easy.

# You should definitely learn this next.

# 4. crosstab()

# Useful for frequency tables.

# Example:

# pd.crosstab(df["Type1"], df["Legendary"])

# # Very useful in EDA.
```
