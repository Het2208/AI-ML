```python
import pandas as pd
```

```python
# Conditional Selection
df = pd.read_csv("pokemon_150.csv")
print(df)
```

```python
# Single Condition
# df[condition]
print(df[df["Weight"] > 90])
print("*"*100)
print(df[df["Height"] < 0.9])
print("*"*100)
print(df[df["Name"] == "Pikachu"])
```

```python
# Multiple Conditions
# Each condition MUST be inside brackets.

# AND condition (&)
# Both conditions must be True.
print(df[(df["Weight"] > 90) & (df["Height"] < 0.9)])

print("*"*100)

# OR condition (|)
# Any one condition can be True.
print(df[(df["Weight"] > 90) | (df["Height"] < 0.5)])

print("*"*100)

# NOT condition (~)
# Reverse condition.
print(df[~(df["Weight"] < 90)])
```

```python
# isin()
# Used when checking multiple values.
print(df[(df["Name"] == "Pikachu") | (df["Name"] == "Mewtwo")]) # Bad way

print("*"*100)

print(df[df["Name"].isin(["Pikachu" , "Mewtwo"])]) # Best way
```

```python
# between() -> check range
print(df[df["Weight"].between(80,90)])
```

```python
# String Filtering

# contains()
print(df[df["Name"].str.contains("P")])

print("*"*100)

# startswith()
print(df[df["Name"].str.startswith("Pi")])

print("*"*100)

# endswith()
print(df[df["Name"].str.endswith("chu")])
```

```python
# Missing Values Filtering

# Find NaN -> isnull()
print(df[df["Type2"].isnull()])

print("*"*100)

# Find non-NaN -> notnull()
print(df[df["Type2"].notnull()])
```

```python
# query()
print(df.query("Height > 90 and Weight < 0.5"))
```

```python
# Selecting Specific Columns After Filtering
print(df[df["Weight"]>90][["Name" , "Weight"]])
```

```python
# Using loc with Conditions
df.loc[df["Height"] < 0.5 , ["Name" , "Height"]] 
```

```python
top_pokemon = df.loc[(df["Type2"].notnull()) & (df["Legendary"] == 1) , ["Name" , "Type1" , "Type2" , "Legendary"]]
print(top_pokemon)
```
