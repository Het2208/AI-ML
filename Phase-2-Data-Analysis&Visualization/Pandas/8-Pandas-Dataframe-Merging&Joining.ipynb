```python
import pandas as pd
```

```python
left_df = pd.DataFrame({"Id" : [101,102,103] , "Name" : ["Het" , "Meet" , "Preet"]})
```

```python
right_df = pd.DataFrame({"Id" : [101,102,103] ,  "Mark" : [90 , 80 , 70]})
```

```python
# PART 1 : merge() is like SQL JOIN.
pd.merge(left_df , right_df , on="Id")
```

```python
# There are 4 main joins.
# Exactly same as SQL joins.
```

```python
# 1. INNER JOIN (Default)
# Keeps only matching rows.
student = pd.DataFrame({"Id" : [101,102,103] , "Name" : ["Het" , "Meet" , "Preet"]})
mark = pd.DataFrame({"Id" : [101,102,104] ,  "Mark" : [90 , 80 , 70]})
pd.merge(student , mark , on="Id" , how="inner")
```

```python
# 2. LEFT JOIN
# Keeps ALL rows from LEFT dataframe.
pd.merge(student , mark , on="Id" , how="left")
```

```python
# 3. RIGHT JOIN
# Keeps ALL rows from RIGHT dataframe.
pd.merge(student , mark , on="Id" , how="right")
```

```python
# 4. OUTER JOIN
# Keeps ALL rows from BOTH tables.
pd.merge(student , mark , on="Id" , how="outer")
```

```python
# PART 2 — Different Column Names
# Sometimes common columns have different names.
student2 = pd.DataFrame({"Id" : [101,102,103] , "Name" : ["Het" , "Meet" , "Preet"]})
mark2 = pd.DataFrame({"StudentId" : [101,102,104] ,  "Mark" : [90 , 80 , 70]})
pd.merge(student2,mark2,left_on="Id",right_on="StudentId")
```

```python
# PART 3 — Multiple Key Columns
# You can merge using multiple columns.
student3 = pd.DataFrame({"Id" : [101,102,103] , "Name" : ["Het" , "Meet" , "Preet"] , "City" : ["Rajkot" , "Baroda" , "Ahmedabad"]})
mark3 = pd.DataFrame({"Id" : [101,102,103] ,  "Mark" : [90 , 80 , 70] , "City" : ["Rajkot" , "Baroda" , " Surat"]})
pd.merge(student3,mark3,on=["Id","City"])
```

```python
# PART 4 — Duplicate Column Names
# Suppose both tables contain "Name" column.
student4 = pd.DataFrame({"Id" : [101,102,103] , "Name" : ["Het" , "Meet" , "Preet"]})
mark4 = pd.DataFrame({"Id" : [101,102,104] ,  "Name" : ["Het" , "Meet" , "Rohan"] , "Mark" : [90,80,70]})
pd.merge(student4,mark4,on="Id")
pd.merge(student4,mark4,on="Id",suffixes=["-Student" , "-Mark"])
```

```python
# PART 5 — indicator=True
# Very important for debugging merges
pd.merge(student,mark,on="Id",how="outer",indicator=True)
```

```python
# PART 6 — Merge on Index
# Normally merge uses columns.
# Can also merge using index.
pd.merge(student,mark,left_index=True,right_index=True)
```

```python
# PART 7 — join() Function
# Simpler version of merge.
# Mostly used with indexes.
student2.join(mark2)
```

```python
pd.merge(student2,mark2,left_on="Id",right_on="StudentId")
```

```python
# PART 8 — concat() ⭐
# Very important.
# Used to STACK dataframes.
# Two types:
# Vertical stacking
# Horizontal stacking
```

```python
# Vertical Concatenation
pd.concat([student,mark])
```

```python
# Horizontal Concatenation
pd.concat([student,mark] , axis=1)
```

```python
pd.concat([student,mark] , ignore_index=True)
```
