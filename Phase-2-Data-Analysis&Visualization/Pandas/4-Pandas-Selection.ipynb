```python
import pandas as pd
```

```python
df = pd.read_csv("pokemon_150.csv")
print(df)
print(df.to_string())
```

```python
# Selection By Column
print(df["Name"].to_string())
# print(df["Height"].to_string())
# print(df["Weight"].to_string())
```

```python
# Multiple Column Selection
print(df[["Name" , "Height" , "Weight"]].to_string())
```

```python
# Selection By Row
print(df.loc[0])
```

```python
df2 = pd.read_csv("pokemon_150.csv", index_col="Name")
# For selection by name
print(df2.loc["Pikachu"])
print("*"*100)
print(df2.loc["Mewtwo" , ["Height" , "Weight"]])
print("*"*100)
print(df2.loc["Moltres":"Mewtwo" , ["Height" , "Weight"]]) # slicing for selection between range
```

```python
# Iteger location
print(df.iloc[0:11]) # second(11) in exclusive
print("*"*100)
print(df.iloc[0:11:2]) # step
print("*"*100)
print(df.iloc[0:11:2 , 0:3]) # 0:3 selects first 3 column index -> 0,1,2 (3 is exclusive)
```

```python
pokemon = input("Enter a name of pokemon : ")
try:
    print(df2.loc[pokemon])
except KeyError:
    print(f"{pokemon} not found")
```

```python
# Selecting rows & columns together
print(df.loc[0:3 , ["Name" , "Legendary"]])
```

```python
# Slicing with iloc properly
print(df.iloc[0:11 , 1:3]) 
# row(index) -> 0-11 & col(index) -> 1-2 
```

```python
# Column selection using dot notation
print(df.Name)
```

```python
# Selecting using list of row indices (not in range but selection of specific row)
print(df.loc[[0,2,3,7]]) 
```
