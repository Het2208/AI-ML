```python
import matplotlib.pyplot as plt
```

```python
# Pie chart shows:
# proportion/percentage of categories
# using circular slices.
```

```python
subjects = ["Maths", "Science", "English"]
marks = [40, 35, 25]

plt.pie(marks)

plt.show()
```

```python
colors = ["red", "blue", "green"]
explode = [0.1 , 0 , 0]
plt.pie(
    marks,
    labels=subjects,
    autopct="%1.1f%%",
    colors = colors,
    explode = explode,
    startangle = 180
)
plt.show()
```
