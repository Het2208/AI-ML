```python
# VIOLIN PLOT

# Violin plot is:
# boxplot + distribution shape

# combined together.
```

```python
# Boxplot shows:
# summary statistics

# Violin plot additionally shows:
# data density/shape
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({"Salary": [25000,30000,35000,40000,45000,500000]})
print(df)
```

```python
sns.violinplot(data=df["Salary"])
plt.show()
```

```python
# Violin width represents:
# where more data points exist

# Wide Area
# many data points concentrated

# Thin Area
# few data points
```

```python
#         Boxplot	         Violin Plot
# Statistical summary	Statistical + distribution
# Simpler	            More detailed
# Better for outliers	Better for shape
# Easier to read	    More informative
```

```python
# CATEGORY VIOLIN PLOT
```

```python
df = pd.DataFrame({"Department" : ["HR" , "HR" , "IT" , "IT"] , "Salary" : [30000 , 35000 , 70000 , 90000]})
print(df)
```

```python
sns.violinplot(x="Department" , y="Salary" , data=df)
plt.plot()
```

```python
sns.violinplot(y="Department" , x="Salary" , data=df)
plt.plot()
```

```python
# INNER PARAMETER

sns.violinplot(
    data=df["Salary"],
    inner="quartile"
)

# Options :
#  Value	            Meaning
#  "box"	          Mini boxplot
#  "quartile"	      Quartile lines
#  "point"	          Individual points
#   None	          Nothing inside
```

```python
# SPLIT VIOLIN

sns.violinplot(
    x="Department",
    y="Salary",
    split=True,
    data=df
)
```

```python
# Use Violin Plot When:
# analyzing distribution shape
# seeing density
# deeper statistical understanding

# Violin Plot Used For
# distribution comparison
# probability density understanding
# advanced EDA
```
