```python
# Seaborn Heatmap

# Heatmap is one of the MOST IMPORTANT visualization techniques in:

# Data Analysis
# Machine Learning
# Correlation analysis
# EDA (Exploratory Data Analysis)

# visualizing numerical intensity using colors
```

```python
# Heatmap means:

# a table where values are represented using colors

# BASIC IDEA OF HEATMAP 

# Heatmap converts:

# numbers → colors
# Example
# Value	Color
# Small	Light
# Large	Dark
# or opposite depending on colormap.
```

```python
# MOST COMMON USE CASE — CORRELATION MATRIX

# Correlation means:
# relationship between numerical columns

# Feature 1	      Feature 2
# Study Hours	    Marks

# If:
# study increases → marks increase

# then:
# positive correlation


# Correlation Values Range  : 
# Value 	Meaning
# 1     	Perfect positive
# 0	        No relation
# -1	    Perfect negative
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "Hours": [1,2,3,4,5],
    "Marks": [20,40,60,80,100],
    "Sleep": [8,7,6,5,4]
})

print(df)
```

```python
corr = df.corr() # relationship between numerical columns
print(corr)
```

```python
# Hours vs Marks
# 1.0

# Perfect positive relation.

# More study:
# → more marks


# Hours vs Sleep
# -1.0

# Perfect negative relation.

# More study:
# → less sleep.
```

```python
sns.heatmap(corr)
plt.show()
```

```python
# COLOR INTERPRETATION 

# Usually:
# dark/red = high positive
# blue = negative
# light = weak relation

# depends on colormap.
```

```python
# annot=True

# Without annotation:
# only colors visible
# Hard to know exact values.

sns.heatmap(corr, annot=True)
plt.plot()


# Displays:
# actual numerical values inside boxes
```

```python
# cmap (Color Map)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.plot()


# cmap          Colors
# "coolwarm"	Blue ↔ Red
# "viridis"  	Purple ↔ Yellow
# "Blues"	    Light ↔ Dark blue
# "Reds"	    Light ↔ Dark red
```

```python
plt.figure(figsize = (6,4))
sns.heatmap(corr, annot=True, cmap="Blues")
plt.plot()
```

```python
sns.heatmap(
    corr,
    annot=True,
    linewidths=2,
    linecolor="black"
)
```

```python
# fmt -> Controls annotation formatting.

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

# show 2 decimal places
```

```python
# cbar -> Controls side color scale bar.

sns.heatmap(corr, annot=True , cbar=False)
```

```python
# Missing Value Heatmap

df = pd.DataFrame({
    "A": [1, None, 3],
    "B": [None, 5, 6]
})

sns.heatmap(df.isna())
```
