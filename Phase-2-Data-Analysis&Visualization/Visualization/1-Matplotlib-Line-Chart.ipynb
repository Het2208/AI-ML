```python
import matplotlib.pyplot as plt
import random
```

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 120, 150, 130]
```

```python
plt.plot(months, sales , linestyle="-." , linewidth=2 , marker="o" , color="red")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.annotate("Highest sales" , xy=["Mar" , 150])
plt.legend(["months","sales"])
plt.savefig("Monthysales" , format="png")
plt.show()
```

```python
months = ["Jan", "Feb", "Mar"]

product_A = [100, 120, 140]
product_B = [90, 110, 150]

plt.plot(months, product_A , label="Product A")
plt.plot(months, product_B , label="Product B")
plt.legend()
plt.show()
```

```python
runs = [random.randint(0,36) for _ in range(1 , 21)]
over = range(1,21)

ax = plt.axes()
ax.set_xlim([0 , 21])
ax.set_ylim([0 , 36])
ax.set_xticks([i for i in range(1,21,1)])
ax.set_yticks([i for i in range(1,36,2)])
ax.grid()
plt.plot(over , runs , marker="*")
plt.show()
```
