```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({"Name": ["Het", "Raj", "Jay"],"Age": [21, np.nan, 22],"Salary": [50000, 45000, np.nan]})

print(df)
```

```python
# PART 2 — isna()
df.isna() # True = Missing value found.
```

```python
# PART 3 — Count Missing Values
df.isna().sum() # Missing values per column

# Why does this work?
# Because:
# True = 1
# False = 0
# So .sum() counts missing values.
```

```python
# PART 4 — Missing Values Percentage
(df.isna().sum() / len(df)) * 100
```

```python
# PART 5 — notna()
# Opposite of isna().
df.notna()

# True = valid value
# False = missing value
```

```python
# PART 6 — Dropping Missing Values (dropna()) 
# Removes rows or columns containing missing values.

# Axis	   Meaning
# axis=0	Rows
# axis=1	Columns
```

```python
# 1. Remove Rows with Missing Values
df.dropna()
```

```python
# 2. Remove Columns with Missing Values
df.dropna(axis=1)
```

```python
# 3. Remove Only if ALL Values Missing
df.dropna(how="all")
```

```python
# 4. Remove Only if ANY Value Missing
df.dropna(how="any")
```

```python
# 5. Threshold (thresh) 
# Keep rows having minimum non-null values.
df.dropna(thresh=2) # Keep rows with at least 2 non-missing values.
```

```python
# PART 7 — Filling Missing Values (fillna()) 
# Instead of deleting data, we fill missing values.
# This is MOST COMMON in Machine Learning.
```

```python
# 1. Fill with Constant Value
df.fillna(0)
```

```python
df.fillna("Unknown")
```

```python
# 2. Fill Specific Column
df["Age"].fillna(0)
```

```python
# 3. Fill Using Mean 
df["Age"].fillna(df["Age"].mean())
```

```python
# 4. Fill Using Median
# Better when outliers exist.
df["Salary"].fillna(df["Salary"].median())
```

```python
# 5. Fill Using Mode
df["Salary"].fillna(df["Salary"].mode()[0])
```

```python
# PART 8 — Forward Fill (ffill)
# Uses previous value.
df.fillna(method="ffill")
```

```python
# PART 9 — Backward Fill (bfill)
# Uses next value.
df.fillna(method="bfill")
```

```python
# PART 10 — Interpolation
# Fills mathematically.
df.interpolate()
```

```python
# PART 11 — Replace Missing Values
# Replace None with NaN
# df.replace(None, np.nan)
```

```python
# PART 12 — Check Rows with Missing Values

# Rows containing NaN
df[df.isna().any(axis=1)]
```

```python
# Rows with NO NaN
df[df.notna().all(axis=1)]
```

```python
# PART 13 — Sorting Missing Values
```

```python
# Put NaN first
df.sort_values(by="Age", na_position="first")
```

```python
# Put NaN last
df.sort_values(by="Age", na_position="last")
```

```python
# PART 14 — Missing Values in GroupBy
print(df)
df.groupby("Age").mean(numeric_only=True)
```
