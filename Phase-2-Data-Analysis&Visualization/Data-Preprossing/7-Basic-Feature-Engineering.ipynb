```python
# Feature Engineering — Complete Beginner-Friendly Explanation

# Feature Engineering is one of the MOST IMPORTANT concepts in Machine Learning.

# Many people think:

# Machine Learning = model building

# But in real-world ML:

# Feature Engineering often matters MORE than model itself.
```

```python
# FIRST UNDERSTAND THE WORD “FEATURE”

# Feature means:

# Input column / input variable

# Example dataset:

# Age	Salary	Experience	Purchased
# 25	40000	2	Yes

# Here:

# Age = feature
# Salary = feature
# Experience = feature

# Target/output:

# Purchased
# WHAT IS FEATURE ENGINEERING?
```

```python
# Feature Engineering means:

# Creating better input features
# from existing data

# to help ML model learn better.
```

```python
# SIMPLE REAL-LIFE ANALOGY 

# Suppose teacher gives raw marks:

# Maths	Science
# 80	90

# Instead of using only raw marks:

# You create:

# Total = Maths + Science
# Average = Total / 2

# These new values may help analysis better.

# That is:

# Feature Engineering
```

```python
# We try to:

# improve data quality
# create useful information
# help model detect patterns
# WHY FEATURE ENGINEERING IS IMPORTANT?

# Because:

# ML models learn from FEATURES.

# Better features:

# → better learning
# → better predictions

# Bad features:

# → poor model performance
# TYPES OF BASIC FEATURE ENGINEERING
```

```python
# Main beginner topics:

# Creating new columns
# Encoding categorical data
# Scaling numerical features
# Extracting information
# Handling dates
# Binning
# Interaction features
# Text features
```

```python
import pandas as pd
```

```python
# 1. Creating New Features

df = pd.DataFrame({
    "Maths": [80, 70, 90],
    "Science": [85, 60, 95]
})

df["Total"] = df["Maths"] + df["Science"]

df["Average"] = df["Total"] / len(df)

print(df)
```

```python
# 2. Extracting Features from Date

df = pd.DataFrame({
    "OrderDate": ["2025-01-10", "2025-03-15"]
})

df["OrderDate"] = pd.to_datetime(df["OrderDate"])

df["Year"] = df["OrderDate"].dt.year

df["Month"] = df["OrderDate"].dt.month

df["Day"] = df["OrderDate"].dt.day

print(df)
```

```python
# 3. Encoding Categorical Features

# Example
# Department
# HR
# IT

# Machine Learning cannot understand text.

# So we use:

# Label Encoding
# One-Hot Encoding

# This is ALSO feature engineering.
```

```python
# 4. Feature Scaling 

# StandardScaler
# MinMaxScaler

# Scaling is feature engineering because:

# we modify features to improve learning.
```

```python
# 5. Binning / Bucketing
# Convert continuous values into groups.

# Age
# 22
# 35
# 60

# Convert into:

# AgeGroup
# Young
# Adult
# Senior

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Age": [22, 35, 60, 48]
})

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0,30,50,100],
    labels=["Young" , "Adult" , "Senior"]
)

print(df)
```

```python
# 6. Interaction Features 

# Combining multiple features.

# Suppose two people:

# Person	Salary	Experience
# A	        50000	5
# B	        50000	1

# Both salaries are same.

# But:

# A earns 10,000 per year experience
# B earns 50,000 per year experience

# Person B is earning much more efficiently.

df = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Salary": [50000, 90000, 120000],
    "Experience": [5, 3, 10]
})

df["Salary_Per_Experience"] = (
    df["Salary"] / df["Experience"]
)

print(df)
```

```python
# 7. Boolean Features ⭐

# Creating True/False columns.

# Why Boolean Features are Useful

# Machine Learning models often learn very well from:

# binary information

# Examples:

# Is customer premium?
# Is age above 18?
# Did student pass?
# Is salary high?
# Is stock price increasing?

# These simple features can contain strong patterns.

df = pd.DataFrame({
    "Age": [15, 22, 35, 12]
})

df["Is_Adult"] = df["Age"] >= 18

print(df)
```

```python
# 8. Handling Missing Values

# Example:

# fill missing values
# create missing indicators

# Why Missing Values are Important

# Most Machine Learning algorithms:

# Cannot handle missing values directly.

# So we must:

# fill them
# remove them
# or create special features from them

import numpy as np

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Salary": [50000, np.nan, 70000, np.nan]
})

df["SalaryMissing"] = df["Salary"].isna()

print(df)
```

```python
# 9. Text Feature Engineering

# Machine Learning models cannot directly understand raw text like humans.
# So we convert text into meaningful features.

# Real-World Applications : 
# Use Case	Example Feature
# Spam detection	message length
# Sentiment analysis	positive words count
# NLP	word count
# Chat analysis	emoji count
# Review analysis	uppercase ratio

df = pd.DataFrame({
    "Name": ["Het", "Meet", "Preet"]
})

df["NameLength"] = df["Name"].str.len()

print(df)
```

```python
# 10. Log Transformation

# Log Transformation is a feature engineering + preprocessing technique used to:
# Reduce skewness in data and make large values more manageable.

# Understanding Logarithm Simply

# Log grows slowly.

# Example:

# Number	log(value)
# 10	     2.30
# 100	     4.60
# 1000	     6.90
# 1000000	 13.81

df = pd.DataFrame({
    "Salary": [30000, 40000, 50000, 10000000]
})

df["LogSalary"] = np.log(df["Salary"])

print(df)
```

```python
# 11. Polynomial Features

# Creating new features using powers or combinations of existing features.

# These features help Machine Learning models learn:

# non-linear (curved) relationships

# instead of only straight-line relationships.

# Real-World Applications
# Field	Example
# Finance	compound growth
# Stock market	price acceleration
# Physics	motion equations
# Population growth	nonlinear increase
# Sales forecasting	seasonal curves

df = pd.DataFrame({
    "Age": [2, 3, 4, 5]
})

df["Age_Squared"] = df["Age"] ** 2

print(df)
```
