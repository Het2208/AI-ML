```python
import pandas as pd
import requests
```

```python
url = "https://archives.nseindia.com/content/equities/EQUITY_L.csv"

symbols = pd.read_csv(url)

symbols[symbols["SYMBOL"] == "SAMMAANCAP"]
```

```python
df = pd.DataFrame()
```

```python
for i in range(10):
    response = requests.get(f"https://api.twelvedata.com/quote?symbol={nseCodes[i]}&exchange=NSE&apikey=42a6157ef30c4ad9a180f7243dd734f7")
    temp_df = pd.json_normalize(response.json())
    df = pd.concat([df, temp_df], ignore_index=True)
```
